```python
from ultralytics import YOLO

# 1. Modeli Seç ve Yükle
model = YOLO("yolo26n.pt")

# 2. Eğitimi Başlat
train_results = model.train(
    data="/content/drive/MyDrive/Colab Notebooks/DataSet/data.yaml",
    epochs=50,
    imgsz=640,
    batch=-1,
    workers=4,
    device=0,
    project="/content/drive/MyDrive/plaka_egitim_sonuclari2",
    name="plaka_modeli_v2",
    save=True,
    plots=True
)
```
